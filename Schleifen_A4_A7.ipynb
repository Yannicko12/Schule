**Aufgabe 4**

a)  
Schulden (Beginn): 1000 <br>
Zinsen pro Jahr: +60 <br>
Rückzahlung: -120 <br>




```python
# Aufgabe 4b
schulden = float(1000)
zinsen = 1.06
rückzahlung = float(120)
counter = 0


while schulden >= 0:
  schulden = (schulden * zinsen) - rückzahlung
  counter += 1
  print("Jahr " + str(counter) + ": " + str(int(schulden)) + "€")

print("Counter: " + str(counter))
```

**Aufgabe 5**

a) <br> 1x1 + 1x2 + 0x4 + 1x8 + 0x16 + 1x32 = 1 + 2 + 8 + 32 = 43 <br>
Die Binärzahl 101011 ergibt 43 im Dezimalsystem. <br>


```python
# Aufgabe 5b

# Eingabe
zahl = int(input("Gib eine Zahl ein, die du in Binär umwandeln möchtest:"))

# Verarbeitung und Ausgabe
while zahl > 0:
  rest = zahl % 2
  print(rest)
  zahl = zahl // 2

print("Lies die Zahl von unten nach oben!")
```

```python
# Aufgabe 5c

# Eingabe
zahl = int(input("Gib eine Zahl ein, die du in Binär umwandeln möchtest:"))
ergebnis = ""

# Verarbeitung
while zahl > 0:
  rest = zahl % 2
  ergebnis += str(rest)
  zahl = zahl // 2
ergebnis = ergebnis[::-1]
# Ausgabe
print(ergebnis)
```

**Aufgabe 6**

Anw. 1) <br>
Es wird 3x untereinander "Hallo" geprintet. <br>
Anw. 2) <br>
Es wird erst 1^2, dann 2^2 und dann 3^2 gerechnet. Also wird 1, 4 und 9 ausgegeben. <br>
Anw. 3) <br>
Weil man bei 0 startet und 3x wiederholt wird, werden die Zahlen 0, 1 und 2 ausgegeben. <br>
Anw. 4) <br>
Diesmal werden die Zahlen von 3 bis 6 ausgegeben, weil man bei 3 startet und vor 7 endet. <br>
Anw. 5) <br>
Es werden die 3 eingetragenen Buchstaben ausgegeben, also "a, b, c".

```python
# Aufgabe 6

for i in [1, 2, 3]:
    print("Hallo")

for i in [1, 2, 3]:
    print(i*i)

for i in range(3):
    print(i)

for i in range(3, 7):
    print(i)

for c in ["a", "b", "c"]:
    print(c)
```
